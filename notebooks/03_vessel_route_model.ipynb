```python
# ============================================================
#  Filter Bulk Carrier Records
# ============================================================

bulk_vessel = vessel[
    vessel["ship_type"] == "Bulk Carrier"
].copy()

bulk_vessel = (
    bulk_vessel
    .sort_values("date")
    .reset_index(drop=True)
)

print("Bulk carrier records:", len(bulk_vessel))

print(
    "Date range:",
    bulk_vessel["date"].min(),
    "to",
    bulk_vessel["date"].max()
)

display(
    bulk_vessel.head()
)
```

```python
# ============================================================
#  Vessel Performance Features
# ============================================================

vessel_features = [
    "route_type",
    "engine_type",
    "maintenance_status",
    "weather_condition",

    "speed_over_ground_knots",
    "engine_power_kw",
    "distance_traveled_nm",
    "draft_meters",
    "cargo_weight_tons",

    "seasonal_impact_score",
    "weekly_voyage_count",
    "average_load_percentage"
]

vessel_target = "turnaround_time_hours"

print("Target:")
print(vessel_target)

print("\nFeatures:")

for feature in vessel_features:
    print("-", feature)
```

```python
# ============================================================
# Prepare Vessel Modeling Data
# ============================================================

vessel_model_data = bulk_vessel[
    ["date", vessel_target] + vessel_features
].copy()

categorical_features = [
    "route_type",
    "engine_type",
    "maintenance_status",
    "weather_condition"
]

for column in categorical_features:
    vessel_model_data[column] = (
        vessel_model_data[column]
        .fillna("Unknown")
        .astype(str)
    )

print(
    "Missing target values:",
    vessel_model_data[vessel_target].isna().sum()
)

display(
    vessel_model_data.head()
)
```

```python
# ============================================================
#  Remove Missing Target Values
# ============================================================

vessel_model_data = (
    vessel_model_data
    .dropna(subset=[vessel_target])
    .reset_index(drop=True)
)

print(
    "Final vessel modeling rows:",
    len(vessel_model_data)
)
```

```python
# ============================================================
#  Chronological Train/Test Split
# ============================================================

vessel_model_data = (
    vessel_model_data
    .sort_values("date")
    .reset_index(drop=True)
)

split_index = int(
    len(vessel_model_data) * 0.80
)

vessel_train = vessel_model_data.iloc[
    :split_index
].copy()

vessel_test = vessel_model_data.iloc[
    split_index:
].copy()

print("TRAIN")
print(
    vessel_train["date"].min(),
    "→",
    vessel_train["date"].max()
)
print("Rows:", len(vessel_train))

print("\nTEST")
print(
    vessel_test["date"].min(),
    "→",
    vessel_test["date"].max()
)
print("Rows:", len(vessel_test))
```

```python
# ============================================================
#  X and y
# ============================================================

X_train_vessel = vessel_train[
    vessel_features
]

y_train_vessel = vessel_train[
    vessel_target
]

X_test_vessel = vessel_test[
    vessel_features
]

y_test_vessel = vessel_test[
    vessel_target
]

print(
    "X_train:",
    X_train_vessel.shape
)

print(
    "X_test:",
    X_test_vessel.shape
)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestRegressor
```

```python
# ============================================================
# Preprocessing Pipeline
# ============================================================

numeric_features_vessel = [
    "speed_over_ground_knots",
    "engine_power_kw",
    "distance_traveled_nm",
    "draft_meters",
    "cargo_weight_tons",
    "seasonal_impact_score",
    "weekly_voyage_count",
    "average_load_percentage"
]

categorical_features_vessel = [
    "route_type",
    "engine_type",
    "maintenance_status",
    "weather_condition"
]

preprocessor_vessel = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            categorical_features_vessel
        ),
        (
            "numeric",
            "passthrough",
            numeric_features_vessel
        )
    ]
)
```

```python
from sklearn.pipeline import Pipeline
# ============================================================
#  Vessel Turnaround Model
# ============================================================

vessel_rf = Pipeline([
    (
        "preprocessor",
        preprocessor_vessel
    ),
    (
        "model",
        RandomForestRegressor(
            n_estimators=300,
            max_depth=8,
            min_samples_leaf=2,
            random_state=42
        )
    )
])

vessel_rf.fit(
    X_train_vessel,
    y_train_vessel
)

print(
    "Vessel turnaround model trained successfully."
)
```

```python
# ============================================================
# Predict Turnaround Time
# ============================================================

y_pred_turnaround = vessel_rf.predict(
    X_test_vessel
)

print("Predictions generated.")

display(
    pd.DataFrame({
        "date": vessel_test["date"].values,
        "actual_turnaround_hours":
            y_test_vessel.values,
        "predicted_turnaround_hours":
            y_pred_turnaround
    }).head(15)
)
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
```

```python
# ============================================================
# Turnaround Model Evaluation
# ============================================================

turnaround_mae = mean_absolute_error(
    y_test_vessel,
    y_pred_turnaround
)

turnaround_rmse = np.sqrt(
    mean_squared_error(
        y_test_vessel,
        y_pred_turnaround
    )
)

print("==========================================")
print("VESSEL TURNAROUND MODEL")
print("==========================================")

print(
    "MAE:",
    round(turnaround_mae, 2),
    "hours"
)

print(
    "RMSE:",
    round(turnaround_rmse, 2),
    "hours"
)
```

```python
# ============================================================
# Turnaround Prediction Graph
# ============================================================
import matplotlib.pyplot as plt
plt.figure(figsize=(12, 6))

plt.plot(
    vessel_test["date"],
    y_test_vessel,
    marker="o",
    label="Actual"
)

plt.plot(
    vessel_test["date"],
    y_pred_turnaround,
    marker="o",
    label="Predicted"
)

plt.xlabel("Date")
plt.ylabel("Turnaround Time (hours)")
plt.title(
    "Bulk Carrier Turnaround Time: Actual vs Predicted"
)

plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
# ============================================================
#  Operational Cost Model
# ============================================================

cost_target = "operational_cost_usd"

print("Target:", cost_target)

print(
    "Missing target values:",
    bulk_vessel[cost_target].isna().sum()
)

display(
    bulk_vessel[cost_target].describe()
)
```

```python
# ============================================================
# Cost Modeling Data
# ============================================================

cost_features = [
    "route_type",
    "engine_type",
    "maintenance_status",
    "weather_condition",

    "speed_over_ground_knots",
    "engine_power_kw",
    "distance_traveled_nm",
    "draft_meters",
    "cargo_weight_tons",

    "seasonal_impact_score",
    "weekly_voyage_count",
    "average_load_percentage"
]

cost_data = bulk_vessel[
    ["date", cost_target] + cost_features
].copy()

categorical_cost = [
    "route_type",
    "engine_type",
    "maintenance_status",
    "weather_condition"
]

for column in categorical_cost:
    cost_data[column] = (
        cost_data[column]
        .fillna("Unknown")
        .astype(str)
    )

cost_data = (
    cost_data
    .dropna(subset=[cost_target])
    .sort_values("date")
    .reset_index(drop=True)
)

print("Cost modeling rows:", len(cost_data))
```

```python
# ============================================================
#  Chronological Split
# ============================================================

cost_split = int(
    len(cost_data) * 0.80
)

cost_train = cost_data.iloc[
    :cost_split
].copy()

cost_test = cost_data.iloc[
    cost_split:
].copy()

print("TRAIN")
print(
    cost_train["date"].min(),
    "→",
    cost_train["date"].max()
)
print("Rows:", len(cost_train))

print("\nTEST")
print(
    cost_test["date"].min(),
    "→",
    cost_test["date"].max()
)
print("Rows:", len(cost_test))
```

```python
# ============================================================
#— X and y
# ============================================================

X_train_cost = cost_train[
    cost_features
]

y_train_cost = cost_train[
    cost_target
]

X_test_cost = cost_test[
    cost_features
]

y_test_cost = cost_test[
    cost_target
]

print("X_train:", X_train_cost.shape)
print("X_test :", X_test_cost.shape)
```

```python
# ============================================================
#  Cost Preprocessing
# ============================================================

numeric_cost = [
    "speed_over_ground_knots",
    "engine_power_kw",
    "distance_traveled_nm",
    "draft_meters",
    "cargo_weight_tons",
    "seasonal_impact_score",
    "weekly_voyage_count",
    "average_load_percentage"
]

categorical_cost = [
    "route_type",
    "engine_type",
    "maintenance_status",
    "weather_condition"
]

preprocessor_cost = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            categorical_cost
        ),
        (
            "numeric",
            "passthrough",
            numeric_cost
        )
    ]
)
```

```python
# ============================================================
# Operational Cost Model
# ============================================================

cost_rf = Pipeline([
    (
        "preprocessor",
        preprocessor_cost
    ),
    (
        "model",
        RandomForestRegressor(
            n_estimators=300,
            max_depth=10,
            min_samples_leaf=2,
            random_state=42
        )
    )
])

cost_rf.fit(
    X_train_cost,
    y_train_cost
)

print(
    "Operational cost model trained successfully."
)
```

```python
# ============================================================
#  Cost Predictions
# ============================================================

y_pred_cost = cost_rf.predict(
    X_test_cost
)

cost_predictions = pd.DataFrame({
    "date": cost_test["date"].values,
    "actual_operational_cost_usd":
        y_test_cost.values,
    "predicted_operational_cost_usd":
        y_pred_cost
})

display(
    cost_predictions.head(15).round(2)
)
```

```python
# ============================================================
#Cost Model Evaluation
# ============================================================

cost_mae = mean_absolute_error(
    y_test_cost,
    y_pred_cost
)

cost_rmse = np.sqrt(
    mean_squared_error(
        y_test_cost,
        y_pred_cost
    )
)

print("==========================================")
print("OPERATIONAL COST MODEL")
print("==========================================")

print(
    "MAE:",
    round(cost_mae, 2),
    "USD/voyage"
)

print(
    "RMSE:",
    round(cost_rmse, 2),
    "USD/voyage"
)
```

```python
# ============================================================
#  Cost Prediction Graph
# ============================================================

plt.figure(figsize=(12, 6))

plt.plot(
    cost_test["date"],
    y_test_cost,
    marker="o",
    label="Actual Cost"
)

plt.plot(
    cost_test["date"],
    y_pred_cost,
    marker="o",
    label="Predicted Cost"
)

plt.xlabel("Date")
plt.ylabel("Operational Cost (USD/voyage)")
plt.title(
    "Bulk Carrier Operational Cost: Actual vs Predicted"
)

plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
from xgboost import XGBRegressor
from sklearn.linear_model import LinearRegression
```

```python
# ============================================================
#  Linear Regression Cost Model
# ============================================================

cost_linear = Pipeline([
    (
        "preprocessor",
        preprocessor_cost
    ),
    (
        "model",
        LinearRegression()
    )
])

cost_linear.fit(
    X_train_cost,
    y_train_cost
)

linear_cost_pred = cost_linear.predict(
    X_test_cost
)

linear_cost_mae = mean_absolute_error(
    y_test_cost,
    linear_cost_pred
)

linear_cost_rmse = np.sqrt(
    mean_squared_error(
        y_test_cost,
        linear_cost_pred
    )
)

print("Linear Regression")
print("MAE :", round(linear_cost_mae, 2))
print("RMSE:", round(linear_cost_rmse, 2))
```

```python
#  XGBoost Cost Model
# ============================================================

cost_xgb = Pipeline([
    (
        "preprocessor",
        preprocessor_cost
    ),
    (
        "model",
        XGBRegressor(
            n_estimators=300,
            max_depth=4,
            learning_rate=0.03,
            subsample=0.8,
            colsample_bytree=0.8,
            objective="reg:squarederror",
            random_state=42
        )
    )
])

cost_xgb.fit(
    X_train_cost,
    y_train_cost
)

xgb_cost_pred = cost_xgb.predict(
    X_test_cost
)

xgb_cost_mae = mean_absolute_error(
    y_test_cost,
    xgb_cost_pred
)

xgb_cost_rmse = np.sqrt(
    mean_squared_error(
        y_test_cost,
        xgb_cost_pred
    )
)

print("XGBoost")
print("MAE :", round(xgb_cost_mae, 2))
print("RMSE:", round(xgb_cost_rmse, 2))
```

```python
# ============================================================
# Naive Cost Baseline
# ============================================================

naive_cost_value = y_train_cost.mean()

naive_cost_pred = np.full(
    len(y_test_cost),
    naive_cost_value
)

naive_cost_mae = mean_absolute_error(
    y_test_cost,
    naive_cost_pred
)

naive_cost_rmse = np.sqrt(
    mean_squared_error(
        y_test_cost,
        naive_cost_pred
    )
)

print("Naive Baseline")
print("MAE :", round(naive_cost_mae, 2))
print("RMSE:", round(naive_cost_rmse, 2))
```

```python
# ============================================================
# Cost Model Comparison
# ============================================================

cost_model_comparison = pd.DataFrame({
    "Model": [
        "Naive Baseline",
        "Linear Regression",
        "Random Forest",
        "XGBoost"
    ],
    "MAE": [
        naive_cost_mae,
        linear_cost_mae,
        cost_mae,
        xgb_cost_mae
    ],
    "RMSE": [
        naive_cost_rmse,
        linear_cost_rmse,
        cost_rmse,
        xgb_cost_rmse
    ]
})

display(
    cost_model_comparison.round(2)
)
```

```python
# ============================================================
#  Best Model Improvement
# ============================================================

best_cost_row = cost_model_comparison.loc[
    cost_model_comparison["MAE"].idxmin()
]

best_cost_model = best_cost_row["Model"]
best_cost_mae = best_cost_row["MAE"]

cost_improvement = (
    (naive_cost_mae - best_cost_mae)
    / naive_cost_mae
) * 100

print(
    "Best cost model:",
    best_cost_model
)

print(
    "MAE improvement over baseline:",
    round(cost_improvement, 2),
    "%"
)
```

```python
# ============================================================
# STEP 32 — FREIGHTWISE ROUTE RECOMMENDATION
# ============================================================

def recommend_route(origin, destination, cargo_tons):

    route = get_route_demand(
        origin,
        destination
    )

    if route is None:
        print(
            "No historical route found for:",
            origin,
            "→",
            destination
        )
        return None

    candidates = vessel_candidates.copy()

    candidates[
        "cargo_difference_pct"
    ] = (
        abs(
            candidates["cargo_weight_tons"]
            - cargo_tons
        )
        / cargo_tons
    ) * 100

    candidates[
        "cargo_fit_score"
    ] = np.clip(
        100 - candidates["cargo_difference_pct"],
        0,
        100
    )

    candidates[
        "vessel_suitability_score"
    ] = (
        0.50
        * candidates["cargo_fit_score"]
        +
        0.30
        * candidates["turnaround_score"]
        +
        0.20
        * candidates["load_score"]
    )

    best = candidates.sort_values(
        "vessel_suitability_score",
        ascending=False
    ).iloc[0]

    return {
        "origin": origin,
        "destination": destination,
        "cargo_tons": cargo_tons,
        "historical_route_quantity_tons":
            route["total_quantity"],
        "historical_shipments":
            route["shipment_count"],
        "demand_score":
            route["demand_score"],
        "demand_share_pct":
            route["demand_share_pct"],
        "recommended_route_type":
            best["route_type"],
        "predicted_turnaround_hours":
            best["predicted_turnaround_hours"],
        "vessel_suitability_score":
            best["vessel_suitability_score"]
    }
```

```python
# ============================================================
# STEP 42 — FINAL FREIGHTWISE DECISION ENGINE
# ============================================================

def recommend_freightwise(
    origin,
    destination,
    cargo_tons
):

    # --------------------------------------------------------
    # 1. ROUTE DEMAND
    # --------------------------------------------------------

    route = get_route_demand(
        origin,
        destination
    )

    if route is None:

        return {
            "status": "NO_ROUTE_DATA",
            "message":
                f"No historical route data for "
                f"{origin} -> {destination}"
        }

    demand_score = (
        float(route["demand_score"]) * 100
    )

    # --------------------------------------------------------
    # 2. VESSEL SUITABILITY
    # --------------------------------------------------------

    candidates = vessel_candidates.copy()

    candidates[
        "cargo_difference_pct"
    ] = (
        abs(
            candidates["cargo_weight_tons"]
            - cargo_tons
        )
        / cargo_tons
    ) * 100

    candidates[
        "cargo_fit_score"
    ] = np.clip(
        100
        - candidates[
            "cargo_difference_pct"
        ],
        0,
        100
    )

    candidates[
        "vessel_suitability_score"
    ] = (
        0.50
        * candidates[
            "cargo_fit_score"
        ]
        +
        0.30
        * candidates[
            "turnaround_score"
        ]
        +
        0.20
        * candidates[
            "load_score"
        ]
    )

    best_vessel = (
        candidates
        .sort_values(
            "vessel_suitability_score",
            ascending=False
        )
        .iloc[0]
    )

    vessel_score = float(
        best_vessel[
            "vessel_suitability_score"
        ]
    )

    # --------------------------------------------------------
    # 3. FREIGHT FORECAST
    # --------------------------------------------------------

    latest = (
        freight_forecast
        .sort_values("date")
        .iloc[-1]
    )

    freight_rate = float(
        latest[
            "predicted_freight_usd_day"
        ]
    )

    freight_score = float(
        latest[
            "freight_attractiveness_score"
        ]
    )

    # --------------------------------------------------------
    # 4. FINAL DECISION SCORE
    # --------------------------------------------------------

    final_score = (
        0.40 * freight_score
        +
        0.30 * demand_score
        +
        0.30 * vessel_score
    )

    # --------------------------------------------------------
    # 5. RECOMMENDATION
    # --------------------------------------------------------

    if final_score >= 70:

        recommendation = "HIGH PRIORITY"

    elif final_score >= 50:

        recommendation = "MEDIUM PRIORITY"

    else:

        recommendation = "LOW PRIORITY"

    # --------------------------------------------------------
    # 6. RETURN
    # --------------------------------------------------------

    return {

        "status": "SUCCESS",

        "origin": origin,

        "destination": destination,

        "cargo_tons": cargo_tons,

        "forecast_date":
            str(
                latest["date"].date()
            ),

        "forecast_freight_usd_day":
            round(
                freight_rate,
                2
            ),

        "freight_score":
            round(
                freight_score,
                2
            ),

        "historical_route_quantity_tons":
            round(
                float(
                    route["total_quantity"]
                ),
                2
            ),

        "historical_shipments":
            int(
                route["shipment_count"]
            ),

        "demand_score":
            round(
                demand_score,
                2
            ),

        "predicted_turnaround_hours":
            round(
                float(
                    best_vessel[
                        "predicted_turnaround_hours"
                    ]
                ),
                2
            ),

        "vessel_suitability_score":
            round(
                vessel_score,
                2
            ),

        "final_decision_score":
            round(
                final_score,
                2
            ),

        "recommendation":
            recommendation,

        "port_constraint_status":
            "NOT_AVAILABLE_IN_CURRENT_DATASET"
    }
```

```python
# ============================================================
# STEP 43 — END-TO-END TEST
# ============================================================

result = recommend_freightwise(
    origin="AUSTRALIA",
    destination="PARADIP SEA",
    cargo_tons=60000
)

for key, value in result.items():
    print(
        f"{key}: {value}"
    )
```
